```python
import numpy as np
from scipy.spatial.distance import pdist, squareform, cdist


cycleMeasurement= np.concatenate([pfData[:, cycleBranches], qfData[:, cycleBranches], voltage[:, cycleNodes]], axis= 1)
cycleMeasurement= cycleMeasurement+ np.random.randn(cycleMeasurement.shape[0], cycleMeasurement.shape[1])*1e-3
print(cycleMeasurement.shape)
# 1. Split the data
train_data = cycleMeasurement[:100, :]
test_data = cycleMeasurement[100:, :]

test_data[200:400, :]+= np.random.randn(200, test_data.shape[1])*0.1

gamma = 0.01 
N_train = train_data.shape[0]
N_test = test_data.shape[0]

# --- TRAINING PHASE ---
# Compute the RBF Kernel for training data
K_train = np.exp(-gamma * squareform(pdist(train_data, metric='sqeuclidean')))

# Center the Training Kernel
one_n = np.ones((N_train, N_train)) / N_train
Kc_train = K_train - one_n @ K_train - K_train @ one_n + one_n @ K_train @ one_n

# Perform Eigenvalue Decomposition
eigenvalues, eigenvectors = np.linalg.eigh(Kc_train)
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# Extract Non-Linear Null Space (KVL Constraints)
tolerance = 1e-10
null_space_indices = np.where(eigenvalues < tolerance)[0]
null_space_eigenvectors = eigenvectors[:, null_space_indices]

# --- TESTING PHASE ---
# Compute Cross-Kernel between Test data and Train data
K_test = np.exp(-gamma * cdist(test_data, train_data, metric='sqeuclidean'))

# Center the Test Kernel STRICTLY using Training statistics
one_m = np.ones((N_test, N_train)) / N_train
Kc_test = K_test - one_m @ K_train - K_test @ one_n + one_m @ K_train @ one_n

# Calculate Anomaly Scores (L_AC) by projecting test data onto the training null space
# Kc_test is (N_test, N_train), null_space_eigenvectors is (N_train, K_null)
projections = Kc_test @ null_space_eigenvectors

# The score is the L2 norm of the projection for each test sample
anomaly_scores = np.linalg.norm(projections, axis=1)

print(f"Extracted {len(null_space_indices)} KVL constraints from {N_train} training samples.")
print(f"Calculated anomaly scores for {N_test} test samples.")
print(f"Baseline test score (Noise Floor) Mean: {np.mean(anomaly_scores):.6e}")
```

```python
G, node_pair2id= graph.from_A_to_G(A[:, :])
G, _= graph.from_A_to_G(A[:, :])
G= G.to_undirected()
import matplotlib.cm as cm
import math


def visualize_cycles_min_overlap(G, node_pair2id, cycles, cycle_importance):
    plt.figure(figsize=(16, 10))
    # 1) link_id -> min importance among cycles containing it
    link_min = {}
    for cyc, imp in zip(cycles, cycle_importance):
        for lid in cyc:
            if lid in link_min:
                link_min[lid] = min(link_min[lid], imp)
            else:
                link_min[lid] = imp

    # 2) Build (u,v) -> link_id (include reversed pair for undirected convenience)
    edge2id = dict(node_pair2id)  # copy
    # Add reverse lookups if not present
    for (u, v), eid in list(node_pair2id.items()):
        if (v, u) not in edge2id:
            edge2id[(v, u)] = eid

    # 3) Map graph edges -> min importance (or None if no cycle)
    edge_vals = {}
    for u, v in G.edges():
        eid = edge2id.get((u, v), None)
        edge_vals[(u, v)] = link_min.get(eid, None)  # None means "not in any cycle"

    # 4) Prepare colormap over existing values
    vals = [v for v in edge_vals.values() if v is not None]
    if len(vals) == 0:
        vmin = 0.0
        vmax = 1.0
        norm = plt.Normalize(vmin, vmax)
    else:
        vmin, vmax = float(min(vals)), float(max(vals))
        # handle degenerate case (all equal) to avoid divide-by-zero
        if math.isclose(vmin, vmax):
            vmin -= 1e-12
            vmax += 1e-12
        norm = plt.Normalize(vmin, vmax)
    cmap = cm.get_cmap("Reds")

    # 5) Draw
    pos = nx.spring_layout(G) # seed
    nx.draw_networkx_nodes(G, pos, node_color="white", edgecolors="black", node_size=100)
    nx.draw_networkx_labels(G, pos, font_size=8)
    # Split edges by having value or not
    edges_in = [(e, edge_vals[e]) for e in edge_vals if edge_vals[e] is not None]
    edges_out = [e for e in edge_vals if edge_vals[e] is None]

    # Draw non-cycle edges (light gray, dashed)
    if edges_out:
        nx.draw_networkx_edges(
            G, pos, edgelist=edges_out, edge_color="lightgray", style="dashed", width=1.5
        )

    # Draw cycle edges colored by min-importance
    if edges_in:
        # Color and width by normalized importance
        colors = [cmap(norm(val)) for (_, val) in edges_in]
        widths = 1.5 # optional thickness
        nx.draw_networkx_edges(
            G, pos, edgelist=[e for (e, _) in edges_in],
            edge_color=colors, width=widths
        )
    nx.draw_networkx_edge_labels(G, pos, edge_labels=node_pair2id, font_size= 8, font_color= "black")

    # Colorbar
    # sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    # sm.set_array([])
    # cbar = plt.colorbar(sm)
    # cbar.set_label("Cycle importance (min over overlaps)")
    # plt.title("Cycle Importance with Overlap = MIN")
    plt.axis("off")
    # plt.tight_layout()
    plt.show()

visualize_cycles_min_overlap(G, node_pair2id, cycleList, np.log10(ratiosStd))
```
