```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

np.random.seed(1)
n_samples = 2000

# Isotropic directions: standard Gaussian then normalize
g = np.random.randn(n_samples, 3)
iso_dir = g / np.linalg.norm(g, axis=1, keepdims=True)

# Anisotropic directions: stretch then normalize
stretch = np.array([3.0, 1.0, 0.3])
g2 = np.random.randn(n_samples, 3) * stretch
aniso_dir = g2 / np.linalg.norm(g2, axis=1, keepdims=True)

# --------- Isotropic plot ---------
fig1 = plt.figure(figsize=(6,7))
ax1 = fig1.add_subplot(111, projection='3d')

ax1.scatter(iso_dir[:, 0], iso_dir[:, 1], iso_dir[:, 2], s=5, alpha=0.4)

# ax1.set_title("Isotropic directions (uniform on sphere)", fontsize=16)
ax1.set_xlabel("x", fontsize=14)
ax1.set_ylabel("y", fontsize=14)
ax1.set_zlabel("z", fontsize=14)

s3= np.sqrt(3)
# Remove tick labels
ax1.set_xticks([-1/s3, 0, 1/s3])
ax1.set_yticks([-1/s3, 0, 1/s3])
ax1.set_zticks([-1/s3, 0, 1/s3])

# fig1.tight_layout()

# --------- Anisotropic plot ---------
fig2 = plt.figure(figsize=(6,30))
ax2 = fig2.add_subplot(111, projection='3d')

ax2.scatter(aniso_dir[:, 0], aniso_dir[:, 1], aniso_dir[:, 2], s=5, alpha=0.4)

# ax2.set_title("Anisotropic directions (non-uniform on sphere)", fontsize=16)
ax2.set_xlabel("x", fontsize=14)
ax2.set_ylabel("y", fontsize=14)
ax2.set_zlabel("z", fontsize=14)

# Remove tick labels
ax2.set_xticks([-1/s3, 0, 1/s3])
ax2.set_yticks([-1/s3, 0, 1/s3])
ax2.set_zticks([-1/s3, 0, 1/s3])

# fig2.tight_layout()


```

```python
n= 10
condNums= []
for i in range(n, 3000):
    noiseVec = np.random.randn(i, n)
    condNums.append(np.linalg.cond(noiseVec))
```

```python
plt.figure(figsize=(7,5))
plt.plot(condNums, color='C1')
plt.yscale("log")
plt.xscale("log")
plt.xticks(fontsize=13)
plt.yticks(fontsize=13)
plt.ylabel(r"$cond(X_{pop})$", fontsize= 14)
plt.xlabel(r"$T_{pop}$", fontsize= 14)
plt.grid(which='both', linestyle='--', linewidth=0.4, alpha=0.4)
```
